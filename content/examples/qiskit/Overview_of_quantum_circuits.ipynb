```python
from qiskit import QuantumCircuit
```

```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure([0], [0])
qc.draw(output="mpl")
```

```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.h(1)
qc.measure([0, 1], [0, 1])
qc.draw(output="mpl")
```

```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.z(0)
qc.h(0)
qc.measure([0], [0])

qc.draw(output="mpl")
```

```python
from qiskit.quantum_info import Statevector
```

```python
qc = QuantumCircuit(1, 1)
qc.x(0)
qc.h(0)
qc.ry(-3.14 / 2, 0)
# qc.z(0)
qc.rz(3.14 / 1.41, 0)
qc.h(0)
```

```python
state = Statevector.from_instruction(qc)
state.draw("bloch") # qsphere, 
```

```python
# Получите результаты измерений
counts = state.sample_counts(1024)
print(f"Результаты: {counts}")
```

```python
from qiskit.visualization import plot_histogram
```

```python
plot_histogram(counts, title='Ideal counts of Ramsey interferometry')
```

```python
from qiskit.visualization import plot_state_city
```

```python
plot_state_city(state)
```
